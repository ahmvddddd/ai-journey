### 1. Key Matrix Operations in NumPy

**A. Transposition (.T)**

- What it is: Transposing a matrix simply means flipping it over its diagonal. The rows become columns, and the columns become rows. If your original matrix has a shape of $(m \times n)$, its transpose will have a shape of $(n \times m)$.
- Relatable Example: If you have a spreadsheet where rows are users and columns are features, transposing it turns it into a layout where rows are features and columns are users.

**B. Reshaping (.reshape())**

- What it is: NumPy allows you to change the dimensions (shape) of an array without altering its underlying data, as long as the total number of elements remains identical.- Relatable Example: Taking a single 12-element vector of monthly sales figures and reshaping it into a $3 \times 4$ matrix representing 3 quarters and 4 regions.

**C. Axis-Wise Aggregations (axis=0 vs. axis=1)**

What it is: When working with 2D matrices, you often want to calculate the sum or mean across rows or columns. This is controlled by the axis parameter in NumPy:

- `axis=0:` Operates vertically down the rows (calculating a metric for each column).
- `axis=1:` Operates horizontally across the columns (calculating a metric for each row).

```python
import numpy as np

# 1. Create a sample matrix: 3 users, 4 features each
# Features: [Age, Income ($k$), Spending Score, Years Active]
user_matrix = np.array([
    [25, 50, 70, 2],
    [32, 85, 40, 5],
    [45, 120, 90, 8]
])

print("--- Original User Matrix ---")
print(user_matrix)
print(f"Shape: {user_matrix.shape}\n")

# 2. Transposition (.T)
# Flips the matrix: rows become columns, columns become rows
transposed_matrix = user_matrix.T
print("--- Transposed Matrix (Shape: {}) ---".format(transposed_matrix.shape))
print(transposed_matrix, "\n")

# 3. Reshaping Data (.reshape)
# Flatten or restructure the 3x4 matrix into a 2x6 matrix
reshaped_matrix = user_matrix.reshape(2, 6)
print("--- Reshaped Matrix (2x6) ---")
print(reshaped_matrix, "\n")

# 4. Axis-Wise Calculations
# Calculate the mean value for each feature (down the rows -> axis=0)
feature_means = np.mean(user_matrix, axis=0)
print(f"Average value per feature (axis=0): {feature_means}")

# Calculate the total score/metric sum for each individual user (across columns -> axis=1)
user_totals = np.sum(user_matrix, axis=1)
print(f"Total metric sum per user (axis=1): {user_totals}")
```

### Guided Question:

When using np.mean(matrix, axis=0) on a dataset matrix, are you calculating the average across individual rows or down individual columns? Explain what the resulting output represents.

### Answer:

When you use `np.mean(matrix, axis=0)`, you are calculating the average down individual columns (collapsing the rows).

**Understanding `axis=0`**

In NumPy, specifying an axis tells the function which axis to eliminate by performing the operation across it:
- `axis=0` operates vertically, moving downwards along the rows for each column.
- `axis=1` operates horizontally, moving across the columns for each row.

**What the Resulting Output Represents**

In a standard dataset matrix where rows represent observations (samples) and columns represent features (variables):
- `np.mean(matrix, axis=0)` computes the mean of each feature across all samples.
- If your matrix has a shape of $(100 \times 5)$—meaning 100 rows and 5 columns—applying `np.mean(matrix, axis=0)` will return a 1D array of shape $(5,)$.
- Each of the 5 values in the output is the arithmetic mean of all 100 entries in that specific column.

For context, if you wanted the average for each individual row instead, you would use axis=1, which would return a shape of $(100,)$ containing 100 values (one average per row).

### Guided Question:

If you have a flat 1D NumPy array containing 24 numbers, what happens if you try to use .`reshape(5, 5)` on it, and why does NumPy raise an error?

### Answer:

If you try to use `.reshape(5, 5)` on a flat 1D NumPy array containing 24 numbers, NumPy will immediately raise a ValueError (specifically, an error stating that it cannot reshape an array of size 24 into shape `(5, 5)`).

**Why Does NumPy Raise an Error?**

The fundamental rule of reshaping in NumPy is that the total number of elements must remain exactly the same before and after the reshape.Your array size: 24 elements.
- The requested shape: $(5 \times 5)$, which requires 25 elements ($5 \text{ rows} \times 5 \text{ columns} = 25$).
- NumPy views reshaping as a way to reorganize how data is arranged, not how much data exists. It cannot magically invent a 25th number to fill out the grid, nor can it leave a slot empty. Every single element from the original array must find a home in the new shape without any leftovers or shortages.

**What Shapes Would Work for 24 Elements?**

To successfully reshape a 1D array of 24 numbers into a 2D matrix, the two dimensions you choose must multiply together to equal 24 (they must be factors of 24).
Valid shapes for a 24-element array include:
- (4, 6) because $4 \times 6 = 24$
- (6, 4) because $6 \times 4 = 24$
- (3, 8) because $3 \times 8 = 24$
- (2, 12) because $2 \times 12 = 24$
- (1, 24) because $1 \times 24 = 24$
