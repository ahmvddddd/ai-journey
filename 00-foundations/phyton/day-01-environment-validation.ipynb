# Day 1 — Environment Validation

This notebook validates the Python/AI-ML development environment established for the AI/ML Engineering Journey.

## Validation goals
- Confirm the active Python interpreter and version.
- Confirm pip is available in the active environment.
- Confirm the core scientific/ML packages import successfully.
- Run small sanity checks with NumPy and Pandas.
- Confirm JupyterLab is installed.

**Expected result:** every validation cell runs successfully without import or assertion errors.

```python
import sys
import platform

print("Python executable:", sys.executable)
print("Python version:", sys.version)
print("Platform:", platform.platform())

assert sys.version_info >= (3, 13), "Python 3.13 or newer is required for this journey."
print("✓ Python version check passed")
```

```python
import subprocess

result = subprocess.run(
    [sys.executable, "-m", "pip", "--version"],
    capture_output=True,
    text=True,
    check=True,
)

print(result.stdout.strip())
print("✓ pip check passed")
```

```python
import numpy as np
import pandas as pd
import scipy
import sklearn
import matplotlib
import seaborn as sns

print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("SciPy:", scipy.__version__)
print("scikit-learn:", sklearn.__version__)
print("Matplotlib:", matplotlib.__version__)
print("Seaborn:", sns.__version__)
print("✓ All core package imports passed")
```

```python
values = np.array([10, 20, 30, 40, 50])
mean_value = values.mean()
total_value = values.sum()

print("NumPy array:", values)
print("Mean:", mean_value)
print("Sum:", total_value)

assert total_value == 150
assert mean_value == 30
print("✓ NumPy sanity check passed")
```

```python
df = pd.DataFrame({
    "product": ["Rice", "Beans", "Oil"],
    "price": [50000, 30000, 45000],
})

print(df)
print("Average price:", df["price"].mean())

assert len(df) == 3
assert df["price"].sum() == 125000
print("✓ Pandas sanity check passed")
```

```python
import jupyterlab

print("JupyterLab:", jupyterlab.__version__)
print("✓ JupyterLab import check passed")
```
