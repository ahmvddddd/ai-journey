 Write a function signature and body for a function named calculate_average_price that accepts a list of floats (prices) and returns a single float representing the average?

```python
def calculate_average_price(prices: list[float]) -> float:
    return sum(prices) / len(prices)
```

The code below incorporates both type hints and a `logging.info()` call inside the function body

```python
import logging

logger = logging.getLogger(__name__)

def process_transaction(amount: float) -> None:
    if amount <= 0:
        raise ValueError("Transaction amount must be greater than zero")

    logging.info("Transaction processed successfully: %.2f", amount)
```

The code below sets the logging level dynamically by reading an environment variable named `LOG_LEVEL` using `os.getenv()`, with a fallback default of "INFO"

```python
import logging
import os

log_level = os.getenv("LOG_LEVEL", "INFO").upper()

logging.basicConfig(level=getattr(logging, log_level, logging.INFO))

logger = logging.getLogger(__name__)

logger.info("Application started successfully")
```
