```python
def calculate_tax(amount, tax_rate = 0.05):
    result = amount * tax_rate
    print(f"{result}")

calculate_tax(100)
calculate_tax(100, 0.10)
```

```python
def get_discount(price, discount_rate = 0.10):
    return price * discount_rate

saved_amount = get_discount(50)
print(saved_amount)
```

```python
def calculate_total_discount(*discounts):
    return sum(discounts)

total_saved = calculate_total_discount(5.0, 12.5, 3.0)
print(total_saved)
```

```python
def log_transaction(**details):
    return details

transaction = log_transaction(item="coffee", cost = 4.5, category = "Food")
for key, value in transaction.items():
    print(f"{key}: {value}")
```

```python
transactions = [12.50, 45.00, 8.25, 100.00, 3.50, 60.00]
large_transactions = []
large_transactions = [transaction for transaction in transactions if transaction > 20.0 ]

print(large_transactions)
```

```python
add_fee = lambda amount: amount + 2.50
print(f"{add_fee(10.00)}")
```

```python

raw_transactions = [
    {"id": 101, "amount": 150.0, "category": "Equipment"},
    {"id": 102, "amount": 12.0, "category": "Software"},
    {"id": 103, "amount": 450.0, "category": "Equipment"},
    {"id": 104, "amount": 8.5, "category": "Supplies"},
]


def process_transactions(transactions, min_amount=0.0, fee_fn=None):
    processed = [
        transaction.copy()
        for transaction in transactions
        if transaction["amount"] >= min_amount
    ]

    if fee_fn is not None:
        for transaction in processed:
            transaction["amount"] = fee_fn(transaction["amount"])
    total_value = sum(
        transaction["amount"] for transaction in processed
    )

    return {
        "processed": processed,
        "total_value": total_value
    }


result = process_transactions(
    raw_transactions,
    min_amount=10.0,
    fee_fn=lambda amt: amt + 2.50
)

print(result)
```
