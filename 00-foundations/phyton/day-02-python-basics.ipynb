```python
tracker = [
    {"name": "Lunch", "cost": 15.0},
    {"name": "Shoes", "cost": 120.0},
    {"name": "Bus", "cost": 5.0}
]
budget = 100.0
total_spent = 0.0
for amount in tracker: total_spent += amount["cost"]
if total_spent > budget:
    amount = total_spent -budget
    print(f"Budget exceeded by ${amount}")
else:
    print("Within budget!")     
```

```python
expenses = []
while True:
    user_choice = input("Choose an action (add / view / exit): ")
    if user_choice == "exit":
        print("Goodbye!")
        break 
    elif user_choice == "add":
        item_name = input("Enter expense name: ")
        amount = float(input("Enter amount: $"))
        user_expenses = {"name": item_name, "amount": amount}
        expenses.append(user_expenses)
        print(expenses)
    elif user_choice == "view":    
        if not expenses:
            print("No expenses recorded yet!")
        else:
            total_spent = 0.0
            for cost in expenses: 
                total_spent += cost["amount"]
                print(f"- {cost['name']}: ${cost['amount']}")
            print(f"Total: {total_spent}")          
```
