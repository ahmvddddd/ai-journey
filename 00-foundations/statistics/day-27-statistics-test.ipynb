### Problem 1: The Skewed Revenue Distribution

An e-commerce analytics platform reports the following metrics for daily user purchase amounts:

- Mean: $145.50
- Median: $42.00
- Mode: $19.99
- Standard Deviation: $210.00

Questions:

1. Is this distribution right-skewed or left-skewed? Explain why based on the relationship between the mean and the median.

2. If you were presenting key business performance metrics to stakeholders to represent a "typical" user purchase, would you recommend reporting the mean or the median? Why?

### Problem 1 Answer:

1. **Is this distribution right-skewed or left-skewed?**
This distribution is right-skewed (also called positively skewed).

**Why:**
In statistics, the relationship between the mean and the median tells us about the shape of the data:

- The median ($42.00) represents the exact middle point where half the purchases are higher and half are lower.
- The mean ($145.50) is pulled significantly higher than the median.

When the mean is much greater than the median, it means the data has a long "tail" on the right side. In e-commerce terms, this happens when most customers make smaller, everyday purchases, but a small group of customers make exceptionally large purchases. Those high-value orders pull the overall average (mean) up, while the median stays grounded where the majority of everyday shoppers actually are.

2. **Would you recommend reporting the mean or the median to represent a "typical" user purchase?**
You should recommend reporting the median ($42.00).

**Why:**
- The mean ($145.50) is heavily distorted by extreme values (the few very large purchases). If you told stakeholders that the "typical" user spends roughly $145, you would be giving them a misleading picture because the vast majority of your customers are actually spending much less (closer to the median of $42 or the mode of $19.99).
- The median gives a true reflection of the middle-of-the-pack customer experience, ensuring your business decisions are based on what regular, everyday shoppers are doing rather than being skewed by a few massive transactions.

### Problem 2: Standard Deviation & Outlier Fences

A dataset of server latency measurements has a first quartile ($Q_1$) of $35.0$ ms and a third quartile ($Q_3$) of $65.0$ ms.

**Questions:**

1. Calculate the Interquartile Range (IQR).
2. Using the Tukey Outlier Fence method ($Q_3 + 1.5 \times \text{IQR}$), what is the upper bound threshold above which any latency value is classified as an outlier?
3. Why is the IQR-based fence more robust to extreme server performance spikes than using mean $\pm$ $3 \times$ standard deviation?

### Problem 2 Answer:

1. **Calculate the Interquartile Range ($\text{IQR}$)**
The Interquartile Range ($\text{IQR}$) measures the middle $50\%$ of the data spread. It is calculated by taking the difference between the third quartile ($Q_3$) and the first quartile ($Q_1$):$$\text{IQR} = Q_3 - Q_1$$Plugging in the values:$$\text{IQR} = 65.0\text{ ms} - 35.0\text{ ms} = 30.0\text{ ms}$$

2. **Upper Bound Threshold (Tukey Outlier Fence)**
Using Tukey's method, the formula for the upper outlier fence is:$$\text{Upper Bound} = Q_3 + (1.5 \times \text{IQR})$$Plugging in our values:$$\text{Upper Bound} = 65.0\text{ ms} + (1.5 \times 30.0\text{ ms})$$$$\text{Upper Bound} = 65.0\text{ ms} + 45.0\text{ ms} = 110.0\text{ ms}$$

Any server latency measurement above $110.0\text{ ms}$ is classified as an outlier.

3. **Why the $\text{IQR}$-Based Fence is More Robust**
The $\text{IQR}$-based method is far more reliable for detecting outliers in server performance data for two main reasons:
- Resistance to Extreme Values: Both $Q_1$ and $Q_3$ rely on percentiles (position in sorted data) rather than summing actual values. Extreme performance spikes (e.g., a momentary lag spike of $5,000\text{ ms}$) do not change $Q_1$ or $Q_3$, keeping the $\text{IQR}$ stable.
- Flaw of Mean and Standard Deviation: The mean and standard deviation are heavily inflated by extreme values. A single huge latency spike inflates both the mean and the standard deviation so much that the threshold ($\text{Mean} + 3 \times \text{SD}$) shifts upward dramatically. This can cause other real outliers to go completely undetected because the threshold was blown out of proportion by the spike itself.

### Problem 3: Correlation Nuances

You analyze two features in a machine learning dataset and find a Pearson correlation coefficient of $r = 0.05$, but a Spearman rank correlation coefficient of $\rho = 0.88$.

**Questions:**

1. What does this massive discrepancy between Pearson and Spearman coefficients tell you about the underlying mathematical relationship between the two variables?

2. Why would feeding these variables into a standard linear regression model without transformation yield misleading feature importance scores?

### Problem 3 Answer:

1. **What the Discrepancy Tells You About the Relationship**
This massive gap between the Pearson ($r = 0.05$) and Spearman ($\rho = 0.88$) coefficients tells you that the relationship between the two variables is strongly monotonic (consistently moving in the same direction) but heavily non-linear (curved).
- Pearson correlation ($r = 0.05$): Measures only straight-line (linear) relationships. A score close to $0$ means that if you try to draw a straight line through the data points, it will be completely flat or useless.
- Spearman correlation ($\rho = 0.88$): Measures monotonic relationships by looking at the ranks (order from smallest to largest) of the data rather than their raw values. A score of $0.88$ means that as one variable increases, the other reliably increases right along with it.

**The Big Picture:** Because Spearman is high and Pearson is near zero, it means the variables are tightly linked, but their pattern looks like a curve (such as an exponential growth curve or a logarithmic curve) rather than a straight diagonal line. When values start small, they might crawl together, but as they grow, they skyrocket, completely breaking the straight-line assumption that Pearson relies on.

2. **Why Linear Regression Yields Misleading Feature Importance Scores**
Feeding these variables directly into a standard linear regression model without any transformation will cause the model to undervalue or completely dismiss the feature, making its feature importance score misleadingly low.The Straight-Line Assumption: Standard linear regression assumes that the relationship between your input feature and the target variable is a straight line ($y = mx + b$).
- The Fitting Failure: Because the actual relationship is curved, a straight regression line cannot fit the data properly. The calculated slope (coefficient) for this feature will end up close to zero because the linear model tries to average out the curve into a flat line.
- The Consequence: When the model evaluates feature importance (often based on the magnitude of these linear coefficients or error reduction), it will look at the terrible straight-line fit and conclude that this feature has little to no predictive power. In reality, the feature is extremely informative—it just needs a transformation (like a log or polynomial transform) or a non-linear model (like a decision tree) to unlock its value.

### Part 2: Practical Python Coding Challenge

Write a complete Python script named week_04_statistics_assessment.py that ingests a synthetic dataset, computes all Week 4 metrics, detects outliers using the IQR method, and calculates a correlation matrix.

**Requirements & Skeleton Code:**

```python
import numpy as np
import pandas as pd

# 1. Setup Data Generation
np.random.seed(42)
n_samples = 1000

# Generate skewed feature (e.g., customer transaction values)
transactions = np.random.exponential(scale=100.0, size=n_samples)
# Generate correlated feature (e.g., time spent on app in minutes)
app_minutes = transactions * 0.05 + np.random.normal(loc=10.0, scale=5.0, size=n_samples)

df = pd.DataFrame({
    "transaction_amount": transactions,
    "app_minutes": app_minutes
})

# ==========================================
# 2. Implement Your Statistical Analysis Below:
# ==========================================

# Compute Mean, Median, and Mode for 'transaction_amount'
mean_txn = df["transaction_amount"].mean()
median_txn = df["transaction_amount"].median()
mode_txn = df["transaction_amount"].mode()[0]

# Compute Sample Variance and Sample Standard Deviation (using Bessel's correction ddof=1)
var_txn = df["transaction_amount"].var(ddof=1)
std_txn = df["transaction_amount"].std(ddof=1)

# Compute Q1 (25th percentile), Q3 (75th percentile), and IQR
q1 = np.percentile(df["transaction_amount"], 25)
q3 = np.percentile(df["transaction_amount"], 75)
iqr = q3 - q1
upper_fence = q3 + 1.5 * iqr

# Count how many rows exceed the upper outlier fence
outlier_count = len(df[df["transaction_amount"] > upper_fence])

# Compute Pearson and Spearman correlation matrices between the two columns
pearson_matrix = df.corr(method="pearson")
spearman_matrix = df.corr(method="spearman")

# Print structured summary report
print("--- Week 4 Statistical Assessment Report ---")
print(f"Mean Transaction:     ${mean_txn:.2f}")
print(f"Median Transaction:   ${median_txn:.2f}")
print(f"Mode Transaction:     ${mode_txn:.2f}")
print(f"Sample Variance:      {var_txn:.2f}")
print(f"Sample Std Dev:       {std_txn:.2f}")
print(f"IQR:                  {iqr:.2f}")
print(f"Outliers Detected:    {outlier_count}")

print("\nPearson Correlation Matrix:")
print(pearson_matrix.round(3))

print("\nSpearman Correlation Matrix:")
print(spearman_matrix.round(3))
```
