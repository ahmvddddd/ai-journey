Matrix multiplication is a foundational building block in machine learning, neural networks, and scientific computing. Unlike element-wise multiplication (`*`), matrix multiplication computes dot products between rows and columns.

To review the dimension matching rule:Multiplying a matrix of shape $(m \times n)$ by a matrix of shape $(n \times p)$ yields a resulting matrix of shape $(m \times p)$.The inner dimensions ($n$) must match, while the outer dimensions ($m$ and $p$) determine the final shape.

### Matrix Multiplication in NumPy


NumPy provides two main ways to execute matrix multiplication:

`np.dot(A, B)`: Traditional matrix dot product.

`A @ B`: The infix operator (introduced in Python 3.5), which is the standard and preferred clean syntax for matrix multiplication.

### Guiding Question:

Suppose we have two 2D arrays representing linear transformation matrices
If we execute `C = A @ B`, what will be the resulting matrix $C$?

```python
import numpy as np

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [7, 8],
    [9, 1],
    [2, 3]
])

C = A @ B

print(C)
print(C.shape)
```

When we execute `C = A @ B`, NumPy performs matrix multiplication. Since `A` has shape `(2, 3)` and `B` has shape `(3, 2)`, the multiplication is valid, and the resulting matrix `C` will have shape `(2, 2)`.

Step 1: Calculate each element
Each element is calculated by multiplying a row of A by a column of B, then adding the products.

\[
C = A @ B
\]

,

- Top-left: \(1(7) + 2(9) + 3(2) = 31\)

For matrix multiplication, the number of columns in the first matrix must equal the number of rows in the second matrix. The resulting matrix takes its rows from the first matrix and its columns from the second matrix.

### Key Matrix Operations in np.linalg

- Transpose  (`A.T`):	Flips rows and columns.
- Determinant (`np.linalg.det(A)`):	Scalar measure of scaling factor and invertibility.
- Inverse (`np.linalg.inv(A)`): Matrix $A^{-1}$ such that $A @ A^{-1} = I$
- Eigenvalues/Vectors	(`np.linalg.eig(A)`):	Decomposes linear transformations into characteristic vectors and scale factors.

### Guiding Question

Suppose we have a simple square system of linear equations:$$\begin{aligned} 2x + y &= 8 \\ x + 3y &= 14 \end{aligned}$$We can represent this system as $A \cdot \vec{x} = \vec{b}$, where $A = \begin{bmatrix} 2 & 1 \\ 1 & 3 \end{bmatrix}$ and $\vec{b} = \begin{bmatrix} 8 \\ 14 \end{bmatrix}$.How would you write the NumPy code using np.linalg.solve(A, b) to find the values of $x$ and $y$?

```python

import numpy as np

A = np.array([
    [2, 1],
    [1, 3]
])

b = np.array([8, 14])

solution = np.linalg.solve(A, b)

print(solution)
print(f"x = {solution[0]}")
print(f"y = {solution[1]}")
```

How it works
- A contains the coefficients of \(x\) and \(y\).
- b contains the right-hand-side constants.
- np.linalg.solve(A, b) finds the vector \(\vec{x}\) satisfying \(A\vec{x}=\vec{b}\).
- solution[0] gives \(x\), and solution[1] gives \(y\).
Why use np.linalg.solve()? It solves the system directly without explicitly calculating the inverse of \(A\), which is generally more efficient and numerically preferabl

### Eigenvalues & Eigenvectors

Another core linear algebra operation in machine learning (especially for dimensionality reduction techniques like PCA) is eigenvalue decomposition.For a square matrix $A$, an eigenvector $\vec{v}$ and its corresponding eigenvalue $\lambda$ satisfy:$$A \cdot \vec{v} = \lambda \cdot \vec{v}$$In NumPy, we compute these simultaneously using `np.linalg.eig(A)`:

```python
eigenvalues, eigenvectors = np.linalg.eig(A)
```

### Matrix Inversion vs. Linear Systems

When working with large linear systems in machine learning (like normal equations in linear regression), computing the explicit inverse matrix $A^{-1}$ using np.linalg.inv() is computationally expensive—$\mathcal{O}(n^3)$ operations—and numerically unstable due to floating-point rounding errors. Instead, we prefer direct solvers like np.linalg.solve() or least-squares approximation:

```python
# Solves Ax = b directly without explicitly computing A^(-1)
x = np.linalg.solve(A, b)

# For overdetermined systems (more equations than unknowns):
x, residuals, rank, s = np.linalg.lstsq(A, b, rcond=None)
```

### Guiding Question

Suppose we have an overdetermined system where $A$ has shape (`100, 5`) (100 observations, 5 features) and $b$ has shape (100,).Why would attempting to use `np.linalg.inv(A)` raise an error, and why is `np.linalg.lstsq(A, b)` the appropriate choice here?

The reason is that A is not a square matrix, and matrix inversion with `np.linalg.inv() `requires a square matrix.
`np.linalg.inv()` strictly requires a square, non-singular matrix (shape $N \times N$). Since $A$ is shape (100, 5), it cannot be inverted directly.np.linalg.lstsq(A, b) solves the least-squares problem for overdetermined systems by finding the coefficient vector $x$ that minimizes the squared error distance $\Vert{}Ax - b\Vert{}^2$.

In other words, it finds the best-fitting solution in the least-squares sense.

```python

import numpy as np

# 100 observations, 5 features
A = np.random.rand(100, 5)

# 100 target values
b = np.random.rand(100)

# Find the least-squares solution
x, residuals, rank, singular_values = np.linalg.lstsq(
    A, b, rcond=None
)

print("Coefficients:", x)
print("Coefficient shape:", x.shape)
print("Residuals:", residuals)
```
