### Challenge 1, Broadcasting Mechanics:

Suppose you have a 2D array A with shape (4, 1) and a 1D array B with shape (3,).

Will the operation C = A + B succeed without raising a ValueError?

If it succeeds, what will be the exact shape of the resulting array C?

1. Yes, `C = A + B` will succeed without raising a ValueError.
2. The resulting array `C` will have the shape `(4, 3)`.
Why? NumPy broadcasting compares dimensions from right to left:
- `A` has shape `(4, 1)`.
- `B` has shape `(3,)`, treated as `(1, 3)` for broadcasting.
- Comparing dimensions: `1` and `3` are compatible, and `4` and `1` are compatible.
The resulting dimensions are `(4, 3)`.

```python
import numpy as np

A = np.array([[1],
              [2],
              [3],
              [4]])  # Shape: (4, 1)

B = np.array([10, 20, 30])  # Shape: (3,)

C = A + B

print(C)
print(C.shape)
```

### Challenge 2, Masking & Filtering

How would you write a single line of `NumPy` code to replace all negative missing-data markers `(-99.0)` with the mean of only the valid positive temperatures in temps?

Suppose we have a 1D temperature dataset:

```python
temps = np.array([18.5, -99.0, 22.0, 30.1, -99.0, 25.4])
```

```python
temps[temps == -99.0] = np.mean(temps[temps > 0])
```

How it works:
- temps > 0 selects only valid positive temperatures.
- np.mean(temps[temps > 0]) calculates their mean: 24.0.
- temps == -99.0 identifies the missing-data markers.
- Assignment replaces those markers with 24.0 in place.

The resulting array is:

```python
[18.5, 24.0, 22.0, 30.1, 24.0, 25.4]
```

### Challenge 3, Axis Aggregations & argsort:

Suppose we have a matrix of student scores across 3 exams:

```python
# Shape: (4, 3) -> 4 students, 3 exams
scores = np.array([
    [85, 90, 78],  # Student 0
    [60, 65, 70],  # Student 1
    [92, 95, 98],  # Student 2
    [75, 80, 85]   # Student 3
])
```

Which axis parameter would you pass to np.mean(scores, axis=...) to calculate the average score for each student across their 3 exams (resulting in a 1D array of 4 averages)?

How would you use np.argsort() on those average student scores to get the indices of the students ordered from lowest average score to highest?

```python
import numpy as np

# 4 students × 3 exams
scores = np.array([
    [85, 90, 78],
    [60, 65, 70],
    [92, 95, 98],
    [75, 80, 85]
])

# Calculate each student's average across exams
averages = np.mean(scores, axis=1)

# Get student indices ordered from lowest average to highest
sorted_indices = np.argsort(averages)

print("Student averages:", averages)
print("Students ranked lowest to highest:", sorted_indices)
```

1. Calculate each student's average
Use axis=1 because you want to calculate the mean across the 3 exam columns for each student.
2. Sort students from lowest average to highest
Pass the averages to np.argsort(). It returns the indices in ascending order of average scores.

Remember:
- axis=0 → calculate down the rows, producing one result per column (exam).
- axis=1 → calculate across the columns, producing one result per row (student).
- np.argsort() → returns the indices that would sort an array in ascending order.

### Challenge 4, Matrix Math & Linear Systems:

Let's test linear system solving in np.linalg.Suppose we want to solve the system $A \cdot \vec{x} = \vec{b}$

b = np.array([9, 8])
What function from np.linalg would you call to solve for vector $\vec{x}$ directly?Why is using that function preferred in numerical computing over calculating np.linalg.inv(A) @ b?

```python
A = np.array([
    [3, 1],
    [1, 2]
])

b = np.array([9, 8])
```

```python
import numpy as np

A = np.array([
    [3, 1],
    [1, 2]
])

b = np.array([9, 8])

x = np.linalg.solve(A, b)

print(x)
```

2. Why is `np.linalg.solve()` preferred?
- More efficient: It solves the system directly without computing the entire inverse matrix.
- Numerically more stable: Direct linear-system solvers generally introduce less unnecessary numerical error than explicitly calculating an inverse and multiplying it by b.
- Better practice: It uses algorithms designed specifically to solve linear systems.
For comparison, avoid this approach when solving a system:

```python
x = np.linalg.solve(A, b)  # Recommended
```

Key takeaway: When you need to solve $A \cdot \vec{x} = \vec{b}$, use `np.linalg.solve(A, b)` rather than calculating the inverse. Note that A must be square and nonsingular for this standard solve operation to work.
