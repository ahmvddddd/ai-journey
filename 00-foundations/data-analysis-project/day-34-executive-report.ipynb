### 1. Data Hygiene & Ingestion Pipeline


Raw enterprise data is inherently noisy. To simulate real-world production environments, the ingestion script purposely injected structural anomalies, including:

- String Padding & Casing Inconsistencies: Leading/trailing whitespaces and mixed casing across text fields (" balogun market, lagos " vs "Wuse Market, Abuja").

- Sentinel Corruption Markers: Negative quantity values (-99) representing logging errors or device dropouts.

- Missing Coordinates: Unrecorded prices and market locations.

**Sanitization Strategy:**

1. Schema Enforcement: Enforced strict dtypes (string, float32, int32, category) during ingestion to optimize memory overhead.

2. Text Normalization: Applied vectorized string methods (.str.strip().str.title()) to standardize market locations, product categories, and vendor names.

3. Contextual Imputation: Missing prices were imputed dynamically using group-wise medians per product category (preserving category pricing tiers), while missing quantities utilized global column medians. Sentinel corruptions were isolated and filtered.


### 2. Exploratory Findings & Regional Performance


**Category Contribution**

Analysis across four core retail sectors—Electronics, Groceries, Apparel, and Home Appliances—revealed distinct revenue drivers:

- High-Value Leaders: Electronics and Home Appliances generated the highest gross merchandise value (GMV) per transaction due to higher baseline unit prices (e.g., Inverters, Generators, and Gas Cookers).

- Volume Drivers: Groceries and Apparel dominated unit turnover frequency, representing consistent, high-velocity cash flow across regional markets.


**Regional Market Hubs**

Evaluating performance across commercial centers highlighted distinct trade patterns:

- Computer Village & Balogun Market (Lagos): Led total GMV generation, driven by high demand for consumer electronics and bulk apparel distribution.

- Wuse Market (Abuja) & Kantin Kwari (Kano): Exhibited robust baseline volumes, with Kano serving as a primary regional hub for textile and staple food commodity distribution.


### 3. Statistical Insights & Distribution Analysis


**Central Tendency & Right-Skewness**

Retail financial data is rarely normally distributed. Our statistical assessment uncovered a pronounced right-skew in transaction pricing and total revenue:

- Price & Revenue Discrepancy: The arithmetic mean transaction revenue was significantly pulled upward by high-ticket wholesale orders and luxury equipment sales, diverging noticeably from the median.

- Business Implication: Reporting mean revenue alone overstates the "typical" consumer purchase. Stakeholders should rely on the median for standard customer behavior benchmarking, while tracking the mean for total revenue auditing.

**Dispersion & Outlier Detection**

Using Bessel's correction (ddof=1) for sample standard deviation and computing the Interquartile Range ($\text{IQR} = Q_3 - Q_1$), we established Tukey Outlier Fences ($Q_3 + 1.5 \times \text{IQR}$):
- High-ticket commercial orders exceeding the upper fence were isolated. Rather than deleting these legitimate wholesale spikes, classifying them separately prevented distortion in standard consumer trend models while preserving enterprise sales visibility.
