# Day 32 — Statistical Analysis (Nigerian Retail Sales Analyzer)

This notebook covers Central Tendency & Skewness, Dispersion & IQR, Tukey Outlier Detection, and Bivariate Correlation Matrices

## Validation goals
- Compute central tendency metrics (mean, median) and distribution skewness for retail prices and total revenue
- Calculate sample standard deviation (using Bessel's correction $\text{ddof=1}$) and Interquartile Range ($\text{IQR}$)
- Implement Tukey Outlier Fences ($Q_3 + 1.5 \times \text{IQR}$) to identify extreme price anomalies
- Construct both Pearson linear and Spearman rank correlation matrices across pricing, quantity, and gross revenue dimensions
- Synthesize statistical insights into structural dataset profiling

With exploratory data analysis complete, Day 32 applies the statistical mechanics mastered in Week 4 directly to our cleaned retail dataset. While EDA shows what happened across regions and categories, statistical analysis reveals the underlying mathematical distribution, spread, outlier thresholds, and bivariate correlations driving our sales figures.

### Statistical Objectives for Day 32

1. Central Tendency & Skewness: Compare mean versus median transaction prices and revenues to quantify distribution skewness caused by high-ticket items (e.g., inverters vs. bags of garri).
2. Dispersion & Robust Spread: Compute sample variance, standard deviation, and the Interquartile Range (IQR).
3. Outlier Detection: Implement Tukey Outlier Fences ($Q_3 + 1.5 \times \text{IQR}$) to isolate anomalous transactions.
4. Correlation Analysis: Evaluate Pearson linear versus Spearman rank correlations between pricing, units purchased, and gross revenue.


### Implementation Script:

```python
import numpy as np
import pandas as pd

# 1. Load Cleaned Dataset
df = pd.read_csv("cleaned_retail_sales.csv", parse_dates=["date"])
df["total_revenue"] = df["price"] * df["quantity"]

# ==========================================
# 2. Central Tendency & Skewness Analysis
# ==========================================
price_mean = df["price"].mean()
price_median = df["price"].median()
price_skew = df["price"].skew()

revenue_mean = df["total_revenue"].mean()
revenue_median = df["total_revenue"].median()
revenue_skew = df["total_revenue"].skew()

print("--- Central Tendency & Distribution Shape ---")
print(f"Price   -> Mean: ₦{price_mean:,.2f} | Median: ₦{price_median:,.2f} | Skewness: {price_skew:.2f}")
print(f"Revenue -> Mean: ₦{revenue_mean:,.2f} | Median: ₦{revenue_median:,.2f} | Skewness: {revenue_skew:.2f}")


# ==========================================
# 3. Dispersion & Robust Spread (IQR)
# ==========================================
price_std = df["price"].std(ddof=1) # Sample standard deviation
price_q1 = np.percentile(df["price"], 25)
price_q3 = np.percentile(df["price"], 75)
price_iqr = price_q3 - price_q1

print("\n--- Dispersion Metrics (Price) ---")
print(f"Sample Standard Deviation: ₦{price_std:,.2f}")
print(f"25th Percentile (Q1): ₦{price_q1:,.2f}")
print(f"75th Percentile (Q3): ₦{price_q3:,.2f}")
print(f"Interquartile Range (IQR): ₦{price_iqr:,.2f}")


# ==========================================
# 4. Tukey Outlier Fence Detection
# ==========================================
upper_fence_price = price_q3 + 1.5 * price_iqr
lower_fence_price = price_q1 - 1.5 * price_iqr

price_outliers = df[(df["price"] > upper_fence_price) | (df["price"] < lower_fence_price)]

print(f"\n--- Outlier Detection (Price) ---")
print(f"Upper Outlier Threshold: ₦{upper_fence_price:,.2f}")
print(f"Detected Outlier Transactions: {len(price_outliers):,} out of {len(df):,}")


# ==========================================
# 5. Bivariate Correlation Analysis
# ==========================================
# Select numerical columns for correlation matrix
numerical_cols = ["price", "quantity", "total_revenue"]

pearson_matrix = df[numerical_cols].corr(method="pearson")
spearman_matrix = df[numerical_cols].corr(method="spearman")

print("\n--- Pearson Correlation Matrix (Linear) ---")
print(pearson_matrix.round(3))

print("\n--- Spearman Correlation Matrix (Monotonic Rank) ---")
print(spearman_matrix.round(3))
```
